```python
# ------------------------------------------------------------
# Task 4: Character counter
# ------------------------------------------------------------
# Ask the user for a sentence. Using a loop, count and print:
#   - the number of vowels (a, e, i, o, u)
#   - the number of digits
#   - the number of spaces
#
# Example:
#   Sentence: CodeStart has 8 sessions
#   Vowels: 7
#   Digits: 1
#   Spaces: 3
#
# Hint: use .lower(), in, .isdigit() and counter variables.
#
# Write your code below:

isSpace = 0
isVowel = 0
isDigit = 0

sen = input("Sentence: ")
sen = sen.lower()

for ch in sen:
    if ch in "aeiou":
        isVowel +=1
    elif ch.isdigit():
        isDigit +=1
    elif ch == " ":
        isSpace +=1
        
print(f"Vowels: {isVowel}\nDigits: {isDigit}\nSpace: {isSpace}")
```

```python
# Exercise 3: Number guessing game
# ------------------------------------------------------------
# Write a game that:
#   1. Picks a secret number from 1 to 10
#   2. Keeps asking for a guess (while loop)
#   3. Says "Too high" or "Too low" (if-elif)
#   4. Stops when the guess is correct (break)
#   5. Prints how many tries it took (counter)
#
# Start with these two lines:
#     import random
#     secret = random.randint(1, 10)
#
# Example:
#   Guess (1-10): 5
#   Too low
#   Guess (1-10): 7
#   Correct! You took 2 tries
#
# Write your code below:

import random as rn 
secret = rn.randint(1, 10)

tries = 0

while True:
    num = input("Guess (1-10): ")
    
    if not num.isdigit():
        print("Numbers only, please!") # can add tries +=1 here if want to count these tries too (optional)
        continue 
    else:
        num = int(num)
        tries += 1

    if num < secret:
        print("Too low")
    elif num > secret:
        print("Too high")
    else:
        print(f"Correct! you took {tries} tries")
        break
```

The general syntax is [start:stop:step]

start: The index to begin at (default is the beginning).
stop: The index to end at (default is the end).
step: How many characters to jump (default is 1, which means moving forward).

When you use [::-1]:
You leave start and stop blank, which means it covers the whole string.
You set step to -1. A negative step tells Python to read the string from right to left (backwards)
